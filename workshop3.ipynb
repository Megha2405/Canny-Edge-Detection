**Canny Edge Detection on a Sample Image**

(i) Implementation of Canny Edge Detection
Objective

The objective is to implement the Canny Edge Detection Algorithm to detect edges in an image.

```python
import cv2
import matplotlib.pyplot as plt


image = cv2.imread("SEC.jpg")


image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)


gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


edges = cv2.Canny(gray, threshold1=100, threshold2=200)


plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(edges, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")

plt.show()
```

**Discussion of Detected Edges**

After applying Canny edge detection:

Major object boundaries become clearly visible.
Fine details and contours are extracted.
Noise is reduced due to Gaussian smoothing.
Strong intensity changes appear as bright edges.

For example:

If the image contains a car, the wheels, doors, windows, and outline become visible.
If the image contains a building, the corners and boundaries are highlighted.

The detected edges are generally thin, continuous, and accurate, making Canny better than simple edge detectors such as Sobel or Prewitt.

**Impact of Different Parameter Settings**

The main parameters in Canny edge detection are:

1. Lower Threshold (threshold1)

Controls weak edge detection.

Low value (e.g., 50)
More edges are detected, including noise.
High value (e.g., 150)
Fewer edges are detected.

```python
edges = cv2.Canny(gray, 100, 300)
```

```python
plt.imshow(edges, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
```

2. **Upper Threshold (threshold2)**

Controls strong edge detection.

Low upper threshold → More edges appear.
High upper threshold → Only strong edges remain.

```python
edges2 = cv2.Canny(gray, 100, 300)
plt.imshow(edges2, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
```
